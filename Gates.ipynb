## Hadamard Gate
The Hadamard gate is a single-qubit quantum logic gate that transforms a definite basis state into an equal superposition of both $|0\rangle$ and $|1\rangle$:
$$H|0\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}}, \qquad H|1\rangle = \frac{|0\rangle - |1\rangle}{\sqrt{2}}$$
$$H = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$
